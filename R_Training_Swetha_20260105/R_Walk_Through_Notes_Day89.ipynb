```r
#handling PDF files in R
library(pdftools)
library(gridExtra)
```

```r
flight_summary<-flights |> group_by(carrier) |> summarise(
  avg_delay=mean(dep_delay,na.rm=TRUE),
  total_flight=n()
)
flight_summary
```

```r
pdf("flight_summary.pdf",width=8,height = 6) #open the file
#writing the conent
plot(
  flight_summary$total_flight,
  flight_summary$avg_delay,
  main="Flights vs average delay",
  xlab="Total flights",
  ylab="Average delay"
)
text(
  flight_summary$total_flight,
  flight_summary$avg_delay,
  labels=flight_summary$carrier,
  pos =4 
)
dev.off() #closing the file
```

```r
library(gridExtra)
pdf("flight_table.pdf",width=10,height = 8) #open the file
grid.table(
  head(flight_summary)
)
dev.off() #close the file
```
