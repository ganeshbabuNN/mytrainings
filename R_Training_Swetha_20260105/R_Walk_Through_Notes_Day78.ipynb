```r
library(tidyverse)
library(nycflights13)
```

```r
# Advanced patterna
#dynamic creating new Column names
col_name <- "new_delay"

flights |> 
  mutate(!!col_name:=arr_delay*2,.keep="used")
```

```r
#column-Wise operations
#across
 flights |> mutate(across(c(dep_delay,arr_delay),~./60),.keep="used")
```

```r
#renaming columns
#dep_delay --> departure_delay
flights |> 
  mutate(departure_delay =dep_delay,.keep="used") |> select(departure_delay)

#replaceing the exising column
flights |> 
  rename(departure_delay =dep_delay) 

#rename the variable where its starts dep_  --> departure_
#if you want prefix to any column
flights |> 
  rename_with(~paste0("Departure_,",.),starts_with("dep_"))
```

```r
#reordering columns
flights |> 
  arrange(arr_delay,dep_delay) #order the value

#reording the variable relocate()
flights |> 
  relocate(arr_delay,dep_delay)

#relative positions
flights |> 
  relocate(arr_delay,.after=arr_time)
```

```r
#type conversion
flights  |> 
  mutate(
    carrier = as.factor(carrier),
    flight=as.character(flight)
  )|> glimpse()

  
  #Bulk type conversion
 flights  |> 
  mutate(across(where(is.numeric),as.double))|> glimpse()
```

```r
# String transformations
flights |> 
  mutate(
    carrier=str_to_lower(carrier),
    .keep="used"
  )
```

```r
#date-time moficiations
flights |> 
  select(year,month,day,time_hour) |> 
  mutate(
    date_str=paste(year,month,day,format(time_hour,"%H"),sep="-"),
    new_time= as.POSIXct(date_str,format="%Y-%m-%d %H")
  )

library(lubridate)
flights |> 
  select(year,month,day,time_hour) |> 
  mutate(
    flight_date=make_date(year,month,day),
    dep_hour=hour(time_hour),
    .keep="used"
  )
```
