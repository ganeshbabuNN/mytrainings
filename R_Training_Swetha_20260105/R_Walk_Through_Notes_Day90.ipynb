```r
#Handling ZIP / Compressed Files in R
flight_data <- flights %>%
  select(
    year,
    month,
    day,
    carrier,
    flight,
    dep_delay
  ) %>%
  slice(1:100)

head(flight_data)

#write csv
write_csv(flight_data,"flight_data.csv")
#create zip file
zip(
  zipfile = "flights.zip",
  files="flight_data.csv"
)

#unzip
unzip(
  "flights.zip",
  list=TRUE
)
```

```r
#unzip
unzip(
  "flights.zip",
  exdir = "unzipped_data"
)

#reading csv without unzipping
read.csv(
  unz("flights.zip","flights.csv")
)
```
