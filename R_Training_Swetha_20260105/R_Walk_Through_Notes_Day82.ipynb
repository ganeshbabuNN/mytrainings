```r
library(tidyverse)
library(nycflights13)
```

```r
# data definations
#creating variable-->mutate()
#conditiona statement --> if_else()
#multi_conditional logic --> case_when()
#modify the existing the variable
#rename the variables- rename()
#data type conversion
flights |> glimpse() |> 
  mutate(
    carrier=as.factor(carrier),
  )
#handling missing ..
##is.na(), coalesce()
#String conversion
```

```r
#combining the columns
flights |> 
  mutate(
    route=paste(origin,dest,sep="-"),.keep="used"
  )

#spliting the columns
flights |> 
  select(time_hour) |> 
  separate(time_hour,into=c("date","time"),sep=" ")

#using across() for bulk transformations
#conditional mutation by group
#joining to defin new data
#recoding values
flights |> 
  mutate(carrier= recode(carrier,
                        "UA"="United",
                       "AA"="American"),.keep="used")

#rowise() definitions
#custom functions 
#trasmute()
#window funcitons
#cummulative functions cumsum()
#any() and all()
#Dynamic column
#row_number()
#nested data nest(list columns) 
nest_data<-flights |> 
  group_by(carrier) |> 
  summarise(data=list(cur_data()))
```

```r
#custom naming with .names
flights |> 
  mutate(
    across(dep_delay:arr_delay,
    mean,
    .names = "mean_{.col}"),
    .keep="used"
  ) |> glimpse()
```

```r
#if_any() and if_all()
#case_match()
#new alternative to recode()
flights |> 
  mutate(carrier_name= case_match(carrier,
                                  "UA"~"United",
                                   "AA"~"American",
                                  .default="Other"),
                                  .keep="used")
```

```r
#vlookup in excel
carrier_lookup <- tibble(
  carrier=c("UA","AA"),
  type=c("Full Service","legacy")
)

flights |> 
  left_join(carrier_lookup,by="carrier") |> 
  select(year,month,day,carrier,type)
```

```r
#chaining definations (layered mutations)
flights |> 
  mutate(delay=arr_delay >0) |> 
  mutate(delay_score=if_else(delay,arr_delay,0))
```

```r
flights |> 
  mutate(delay=arr_delay >0,
         delay_score=if_else(delay,arr_delay,0),.keep="used")
```

```r
#with() is base R functions
with (flights,distance/air_time)
```
