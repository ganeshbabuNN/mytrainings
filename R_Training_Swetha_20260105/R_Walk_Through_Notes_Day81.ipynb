```r
#df-in and df-out
top_n_delay <- function(df,n=3){
  df |> 
    arrange(desc(arr_delay)) |> 
    slice_head(n=n)
}

flights |> 
  group_by(carrier) |> 
  group_modify(~top_n_delay(.x))
```

```r
flights |> 
    group_by(carrier) |> 
    summarise(avg=mean(arr_delay,na.rm=TRUE))
```

```r
#dynamic column using tidy evlautions
#programming with {{}}
multi_mean <- function(data,group_col,val_col){
  data |> 
    group_by({{group_col}}) |> 
    summarise(avg=mean({{val_col}},na.rm=TRUE))
}

multi_mean(flights,carrier,arr_delay)

#ASSIGNMENT : to passe the multiple parameter for group_col.
```

```r
# ... variable argument
multi_mean <- function(data,...){
  data |> 
    summarise(across(c(...),~mean(.x,na.rm=TRUE)))
}

#creating function passing dataframe varibales.
multi_mean(flights,dep_delay,arr_delay)
#flights - dataframe.
#dep_delay - not a df, attribute
#arr_delay - not a df , attribute
```
